```python
# Question 5
a=np.random.randint(1,91,(5,5));
print(a)
print()
b=a.mean(axis=1, dtype=int)
print("Mean of column:",b)
print()
c=a.mean(axis=0,dtype=int)
print("Mean of row:",c)
print()
d=a.max(axis=1)
print("Max in row:",d)
print()
e=a.min(axis=0)
print("MIN in column:",e)
print()
print("converted to negative:")
for i in range(5):
  for j in range(5):
    if(a[i][j]%2==0):
        a[i][j]=-1;

print(a)
print()
f=a.sum(axis=0)
print("sum of row:",f)
g=a.sum(axis=1)
print()
print("sum of column:",g)
count=0
for i in range(5):
  for j in range(5):
    if(a[i][j]<0):
      count=count+1;
print("Count of negative:",count)
```
