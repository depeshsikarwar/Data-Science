```python
lst=[1,2,3,"hello",'c',[2,4]]
print(lst)
# append single element
lst.append(1)
print(lst)

# entend() for multiple element && work on list,dictionary,set,etc;
lst.extend([2,3,4])
print(lst)

# insert add at some position

# lst.insert(index,value) single value
lst.insert(2,[2,3,4])
print(lst)

# pop()
lst.pop()
print(lst)

# remove() it take a value and remove it from list first occurance
lst.remove(3);
print(lst)

# index :return index
print(lst.index(2))

# count()
print(lst.count(1))


# reverse()
lst.reverse()
print(lst)

# Sort
lst1=[1,2,3,4,5,-1,0,2]
lst1.sort()
print(lst1)
```

```python
lst=[1,-2,19,-5,10,7,8,8]
print(sum(lst))
print(min(lst))
print(max(lst))
print(lst)
print(sorted(lst))
# sort make change in a list
# sorted return a new list
```

```python
# TOUPLE
tp=(10,6,2,89,-2,-12)
print(tp)
# we can use those function that return something
print(tuple(sorted(tp)))
# sorted function return list
print(sum(tp))
print(min(tp))
print(max(tp))
print(tp.count(2))
print(tp.index(2))
print(len(tp))
```
