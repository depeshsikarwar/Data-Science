```python
a=input("Enter the string: ")
n=len(a)
i=0
j=n-1
while i<j:
  if(a[i]==a[j]):
    break
  i=i+1;
  j=j-1;

if(i<j):
  print("Palindrome")
else:
  print("Not Palindrome")
```
