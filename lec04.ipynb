```python
a=np.random.randint(1,15,8)
e=np.random.randint(1,15,8)
print(a)
print(e)
b=a.reshape(8,1)
print(b)
c=b.transpose()
print(c)
d=np.concatenate((a.reshape(-1, 1),e.reshape(-1, 1)),axis=1)
d=np.concatenate((a,e),axis=0)
print(d)
```

```python
a=np.random.randint(1,15,8)
b=np.random.randint(1,15,8)
c=np.where(b>a,b,a)
d=np.where(a>b,b,a)
e=np.where(a==b,b,a)
f=np.where(b>2,b,a)
# apply on each value
print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```
