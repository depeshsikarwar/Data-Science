```python
class Student:
    def __init__(self,name,age):
      self.name=name
      self.age=age
    def display(self):
      print("Name:",self.name)
      print("Age:",self.age)
a=Student('john',20)
a.display()
```

```python
class person:
    def __init__(self, h, w, age):
        self.height = h
        self.weight = w
        self.age = age

class Rahul(person):
    def __init__(self, g, name, b,h,w,age):
        super().__init__(h,w,age)
        self.gender = g
        self.name = name
        self.bank_balance = b

a = Rahul('male', 'Rahul', 600000,5,87,45)

print(a.name)
print(a.bank_balance)
print(a.gender)
print(a.height)
print(a.weight)
```

```python
def sum(a,b):
  return a+b
def sum(a,b,c):
  return a+b+c
def sum(a,b,c,d):
  return a+b+c+d
print(sum(1,2))
print(sum(1,2,3))
print(sum(1,2,3,4))
```

```python
class complex:
  def __init__(self,x,y):
    self.x=x
    self.y=y
  def __add__(self,other):
    a=self.x+other.x
    b=self.y+other.y
    result=complex(a,b)
    return result
  def display(self):
    print(self.x,'+i',self.y)
a=complex(1,2)
b=complex(3,4)
c3=a+b
c3.display()
```
