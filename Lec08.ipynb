```python
import pandas as pd
```

```python
df1=pd.DataFrame({
    'id':[1,2,3,4,5],
    'name':['a','b','c','d','e'],
    'age':[20,21,22,23,24],
    'city':['a1','a2','a3','a4','a5']
})


df2=pd.DataFrame({
    'id':[3,4,5,6,7],
    'name':['f','g','h','i','j'],
    'age':[25,26,27,28,29],
    'city':['f1','f2','f3','f4','f5']
})
# inner join
# df3=pd.merge(df1,df2,on='id',how='inner',suffixes=('_table1','_table2'))
df3=pd.merge(df1,df2,on='id',how='inner')
df3.rename(columns={'name_x':'table1_name','name_y':'table1_name'},inplace=True)
# print(df3)
# left join
df4=pd.merge(df1,df2,on='id',how='left',suffixes=('_table1','_table2'))
# print(df4)
# right join
df5=pd.merge(df1,df2,on='id',how='right')
# print(df5)
# outer join
df6=pd.merge(df1,df2,on='id',how='outer')
display(df6)

# cross join
df7=pd.merge(df1,df2,how='cross')
# display(df7)

# Concat //up and right join or stake join
df8=pd.concat([df1,df2])
# display(df8)


```

```python
# DROP DUPLICATES.
df1=pd.DataFrame({
    'id':[1,2,3,4,5],
    'name':['a','b','c','d','e'],
    'age':[20,21,22,23,24],
    'city':['a1','a2','a3','a4','a5']
})


df2=pd.DataFrame({
    'id':[3,4,5,6,7],
    'name':['f','g','h','i','j'],
    'age':[25,26,27,28,29],
    'city':['f1','f2','f3','f4','f5']
})

df3=pd.concat([df1,df2])
df3=df3.drop_duplicates(subset=['id'],keep='last',inplace=True)
display(df3)
```
