```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
# linr chart
x=np.linspace(1,10,100)
y=np.sin(x)
plt.plot(x,y)
```

```python
x=[1,2,3,4,5]
y=[1,4,9,10,15]
plt.figure(figsize=[5,5])
plt.plot(x,y)

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line Chart')
plt.tight_layout()
plt.show()
```

```python
subjects=['Math','Physics','CS','English']
scores=[88,96,34,67]
plt.bar(subjects,scores,color='red',width=0.5,edgecolor='black')
# plt.bar(scores,subjects,color='red',width=0.5,edgecolor='black')
plt.xticks(rotation=45)
plt.xlabel('Subjects')
plt.ylabel('Scores')
plt.title('Bar Chart')
plt.tight_layout()
plt.show()
```

```python
data=np.random.normal(60,10,500)
plt.hist(data,bins=20,edgecolor='black')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.show()
```

```python
subject=['math','science','cs','English']
scores=[30,40,10,20]
plt.pie(scores,labels=subject,autopct='%1.1f%%',shadow=True,startangle=50)
plt.xlabel('Subjects')
plt.ylabel('Scores')
plt.title('Pie Chart')
plt.show()
```

```python
batch1=np.random.randint(10,70,100)
plt.boxplot(batch1,labels=['batch1'])
plt.tight_layout()
plt.show()
```

```python
df=[[1,2,3],[4,5,6],[7,9,8]]
plt.imshow(df,cmap='hot')
plt.colorbar()
plt.show()
```

```python
# line plot through seaborn
df=pd.read_csv("student_data.csv")

# sns.set(style='darkgrid')
sns.lineplot(data=df,x='study_hours',y='exam_score')
plt.tight_layout()
plt.title("Line Plot")
plt.show()
print(df.head())
```

```python
df=pd.read_csv("student_data.csv")
sns.barplot(data=df,x='sleep_hours',y='exam_score')
plt.tight_layout()
plt.title("Hist Plot")
plt.show()
print(df.head())
```

```python
df=pd.read_csv("student_data.csv")
sns.histplot(data=df,x='exam_score')
plt.tight_layout()
plt.title("Hist Plot")
plt.show()
print(df.head())
```

```python
data = np.random.normal(60, 10, 500)
index = np.arange(500)   # 0 to 499

sns.scatterplot(x=index, y=data)

plt.xlabel("Index")
plt.ylabel("Data")
plt.title("Scatter Plot")
plt.show()
```

```python
marks = [40, 45, 50, 55, 60, 65, 70, 75, 80, 120]

sns.boxplot(marks)

plt.title("Box Plot")
plt.show()
```

```python
data = [
    [10,20,30],
    [40,50,60],
    [70,80,90]
]

sns.heatmap(data)

plt.title("Heatmap")
plt.show()
```
